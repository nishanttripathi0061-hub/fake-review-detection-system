```python
import pandas as pd
import os

file_path = "/content/drive/MyDrive/fake reviews dataset.csv"

print("File exists:", os.path.exists(file_path))

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
# 1. Dataset dimensions
print("Dataset shape:", df.shape)

# 2. Column names and data types
print("\nDataset information:")
df.info()

# 3. Missing values
print("\nMissing values:")
print(df.isnull().sum())

# 4. Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 5. Label distribution
print("\nLabel distribution:")
print(df["label"].value_counts())

# 6. Sample reviews
print("\nSample reviews:")
print(df[["label", "text_"]].head(5).to_string(index=False))
```

```python

# Label distribution: count and percentage
print("Label counts:")
print(df["label"].value_counts())

print("\nLabel percentages:")
print((df["label"].value_counts(normalize=True) * 100).round(2))

# Review length analysis
df["text_length"] = df["text_"].astype(str).str.len()

print("\nReview length statistics:")
print(df["text_length"].describe().round(2))

# Shortest and longest reviews
print("\nShortest reviews:")
print(df.nsmallest(3, "text_length")[["label", "text_"]].to_string(index=False))

print("\nLongest reviews:")
print(df.nlargest(3, "text_length")[["label", "text_"]].to_string(index=False))

# Duplicate review text
print("\nDuplicate review texts:", df["text_"].duplicated().sum())
```

```python

import re
import html

def clean_text(text):
    text = html.unescape(str(text))
    text = re.sub(r"<[^>]+>", " ", text)  # Remove HTML tags
    text = re.sub(r"\s+", " ", text)      # Normalize whitespace
    return text.strip()

df["clean_text"] = df["text_"].apply(clean_text)

print("Original review:")
print(df["text_"].iloc[0][:500])

print("\nCleaned review:")
print(df["clean_text"].iloc[0][:500])

print("\nEmpty cleaned reviews:", (df["clean_text"].str.len() == 0).sum())
```

```python
duplicate_mask = df["clean_text"].duplicated(keep=False)

print("Rows in duplicate-text groups:", duplicate_mask.sum())

label_counts = (
    df[duplicate_mask]
    .groupby("clean_text")["label"]
    .nunique()
)

print(
    "Duplicate text groups with conflicting labels:",
    (label_counts > 1).sum()
)
```

```python
import re
import html

def clean_text(text):
    text = html.unescape(str(text))
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["clean_text"] = df["text_"].apply(clean_text)

print("Cleaned text column created!")
print(df.columns.tolist())
```

```python
raw_text = df["text_"].astype(str)

html_mask = raw_text.str.contains(
    r"<[^>]+>", regex=True, na=False
)

whitespace_mask = raw_text.str.contains(
    r" {2,}|\t|\n|\r", regex=True, na=False
)

analysis_df = df.copy()
analysis_df["has_html"] = html_mask
analysis_df["has_extra_whitespace"] = whitespace_mask

print(
    analysis_df.groupby("label")[
        ["has_html", "has_extra_whitespace"]
    ].mean().mul(100).round(2)
)
```
